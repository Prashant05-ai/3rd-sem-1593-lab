```python
# combine and merge the dataset
# student management System
#     student_id,name,Department
#examination system
#     student_id,internal,final
#attendence system
#     student_id,attendence
```

```python
#create the dataframe
s_info = pd.DataFrame({
    "Student_id":["S101","S102","S103","S104"],
    "Name":["Arjun","Lalit","Raj","Priya"],
    "Department":["CSE","DS","CSE","DS"]
})
print(s_info)
```

```python
# DataFrame for Examination System
exam = pd.DataFrame({
    "Student_id":["S101","S102","S103","S105"],
    "Internal":[82,75,90,66],
    "Final":[79,56,84,76]
})
```

```python
# Merge - join
# Types of join
# 1. inner join - intersection
# 2. outer join - union
# 3. left join  - all details in left_table + common info in right table
# 4. right join - all details in right_table + common info in left table
```

```python
# inner join
inner_join=pd.merge(s_info,exam, on="Student_id",how="inner")
inner_join
```

```python
#outer join
outer_join=pd.merge(s_info,exam, on="Student_id",how="outer")
outer_join
```

```python
# left join
left_join=pd.merge(s_info,exam, on="Student_id",how="left")
left_join
```

```python
# right join
right_join=pd.merge(s_info,exam, on="Student_id",how="right")
right_join
```

```python
s1=pd.DataFrame({
    "Student_id":["S101","S102","S103"],
    "Name":["Anu","Ravi","Neha"]
})
exam1=pd.DataFrame({
    "id":["S101","S102","S103"],
    "marks":[88,72,95]
})

# Merge
pd.merge(s1,exam1, left_on="Student_id",right_on="id")
```
