```python
# CSV format
df.to_csv("student_csvoutput.csv", index=False)
student_csv = pd.read_csv("student_csvoutput.csv")
print("CSV Output:")
print(student_csv.head())
```

```python
# Excel format
df.to_excel("student_exceloutput.xlsx", index=False)
student_excel = pd.read_excel("student_exceloutput.xlsx")
print("\nExcel Output:")
print(student_excel.head())
```

```python
# JSON format
df.to_json("student_jsonoutput.json", orient="records", indent=4)
student_json = pd.read_json("student_jsonoutput.json")
print("\nJSON Output:")
print(student_json.head())
```

```python
# Parquet format
df.to_parquet("student_parquetoutput.parquet", index=False)
student_parquet = pd.read_parquet("student_parquetoutput.parquet")
print("\nParquet Output:")
print(student_parquet.head())
```
