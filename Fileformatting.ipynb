```python
CSE=pd.DataFrame({
    "Name":["Ajay","Bansal"],
    "Marks":[90,93]
})
ECE = pd.DataFrame({
    "Name":["Anish","Karan"],
    "Marks":[89,88]
})
IT=pd.DataFrame(
    {"Name":["Harshita","Jhansi"],
     "Marks":[88,90]})
with pd.ExcelWriter("student_deaprtment.xlsx") as f1:
  CSE.to_excel(f1,sheet_name="CSE",index=False)
  ECE.to_excel(f1,sheet_name="ECE",index=False)
  IT.to_excel(f1,sheet_name="IT",index=False)
```
