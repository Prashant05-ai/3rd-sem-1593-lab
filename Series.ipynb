###Q29. Write a short Pandas program that creates a labelled Series for four products and sales, displays its values and index, selects one label, filters sales above 500, checks whether a label exists, and converts the Series to a dictionary

```python
import pandas as pd

products = pd.Series(
    [300, 650, 450, 800],
    index=['Pen', 'Notebook', 'Bag', 'Laptop']
)
print("Values:")
print(products.values)
print("\nIndex:")
print(products.index)
print("\nNotebook sales:")
print(products['Notebook'])
print("\nSales above 500:")
print(products[products > 500])
print("\nIs Bag present?")
print('Bag' in products.index)
print("\nDictionary:")
print(products.to_dict())
```
