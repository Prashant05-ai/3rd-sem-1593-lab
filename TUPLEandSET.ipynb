#Tuple
## ordered collections of items()
## one main difference with is tuple is immutable

```python
# Create a tuple
student = (101,"sanjiv","CSE",8.9)
print(student)

# tuples using index
print("student infornmation in first location :",student[0])
print("student information in second location :",student[1])

#negative index
print("example for negative index :",student[-2])
```

```python
# length of tuple
subjects = ("DSA","DSV","OS","python","machine learning")
print("length of the tuple :",len(subjects))
```

```python
#slicing
subjects = ("DSA","DSV","staticis","sql","machine learning")
print(subjects[1:3])
print(subjects[:5])
print(subjects[4:])
```

```python
# tuple immutable
students_info = ("chirag",21,"CSE",9.7)
students_info[3] = 9.1
print(students_info)
```

```python
# Nested tuple
student = (101,"rahul",(89,97,88))
print(student)
print(student[2])
print(student[2][1])
```

```python
# count function
marks = (75,88,78,75,92,90,98,90)
print("times of occurance:",marks.count(75))
```

```python
# find the element present in tuple or not
subjects  = ("dsa","python","dsv","sql","machine learning","staticis")
print("python" in subjects)
print("java" in subjects)
```

# SET
## unordered collection of items
## set is mutable
## allows the different data types
## it doesn't support for index
## it gives output with uniques (no duplication)


```python
# symbol for set {}
# create a set with similar items
marks  = {75,89,78,55,89,75,66,55}
print(marks)
```

```python
# create a empty set
a = {}
a = set()
# check its type
print(type(a))
```

```python
# set doesn't support the index
students = {"Amit","riya","priya"}
print(students[1])
```

```python
# add an item in the set
students = {"amit","kiran","chirag"}
students.add("priya")
print(students)
# to add multiple values or items
students.update(["ram","chitra","vani","devi"])
print(students)
```

```python
# remove an item from the set
students = {"amit","riya","priya"}
students.remove("priya")
print(students)

students.discard("amit")
print(students)
```

```python
students = {"amit","riya","priya"}
students.remove("rahul")
print(students)
```

```python
students = {"amit","riya","priya"}
students.discard("rhaul")
print(students)
```

```python
# cretae a set and clear the list if items
students_info = {"amit","riya","priya"}
print(students_info)
students_info.clear()
print(students_info)
```

```python
marks = {99,88,78,69}
del marks
print(marks)
```

```python
# set operations
python_students = {"riya","priya","mona","rahul","neha"}
java_students = {"amit","manish","mona","dilip","manish","neha"}

# for union operation by using set
result = python_students.union(java_students)
print("union operations : ",result)

# for intersection operation by using set
result_intersection = python_students.intersection(java_students)
print("result after intersection : ",result_intersection)

# to find difference on set
result_differnece = python_students.difference(java_students)
print("result after difference : ",result_differnece)
```

```python
# convert list to set
student_id = [101,102,103,104,105,104,103]
unique_student = set(student_id)
print(student_id)
print(unique_student)
unique_roll_no=list(unique_student)
print(unique_roll_no)
```
